```python
from tensorflow.keras.models import load_model
model = load_model(r"D:\Knee\kneemodel.keras")
```

```python
from tensorflow.keras.preprocessing import image
import matplotlib.pyplot as plt
import numpy as np
# Function to make predictions on new images
def load_and_preprocess_image(img_path):
    img = image.load_img(img_path, target_size=(150, 150))
    img_array = image.img_to_array(img) / 255.0  # Normalize pixel values
    img_array = np.expand_dims(img_array, axis=0)  # Add batch dimension
    return img_array
```

```python
# Example usage for prediction
img_path = r'D:\Knee\test\2Mild\MildG2 (2).png'# Replace with image path
processed_image = load_and_preprocess_image(img_path)
predictions = model.predict(processed_image)

# Get the predicted class
predicted_class_index = np.argmax(predictions[0])  # Get the index of the highest probability
categories = ['0Normal', '1Doubtful', '2Mild', '3Moderate','4Severe']
predicted_class = categories[predicted_class_index]
print(f'Predicted Class: {predicted_class}')

# Plot the image and the predicted class
plt.imshow(image.load_img(img_path, target_size=(150, 180)))  # Load the original image for plotting
plt.title(f'Predicted Class: {predicted_class}')
plt.axis('off')  # Hide axes
plt.show()
```
